# **Pairs Trading Mean-Reversion Model to exploit the Russell 2000 (IWM) vs. S&P 500 (SPY) divergence**

This notebook implements a quantitative **pairs trading strategy** built on the statistical framework of cointegration and mean reversion.

While the S&P 500 ($SPY$, tracking large-cap equities) and the Russell 2000 ($IWM$, tracking small-cap equities) broadly share macroeconomic drivers, they frequently experience localized structural trends or liquidity shifts. This model identifies when the historical price ratio between the two indices stretches to an extreme statistical threshold, signaling an entry to **short the outperformer** and **long the underperformer** under the expectation that the spread will revert to its long-term historical mean.

### Key Components

* **Cointegration Testing:** Utilizes the Engle-Granger two-step method to confirm a statistically stable, long-term relationship between the two assets rather than a temporary correlation.
* **Spread & Z-Score Calculation:** Models the price spread dynamically and converts it into a rolling $Z$-score to identify entry ($Z \ge \pm2.0$) and exit ($Z = 0$) thresholds.
* **Vectorized Backtesting:** Evaluates historical strategy performance, accounting for transactional friction and calculating risk metrics like the Sharpe Ratio and Maximum Drawdown.
